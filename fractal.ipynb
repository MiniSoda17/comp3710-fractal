```python
import matplotlib.pyplot as plt
import torch

initial_triangle = torch.tensor([[0.0, 0.0], [1.0, 0.0], [0.5, 0.866]]) 
point = torch.tensor([0.5, 0.5])

points_holder = torch.zeros((1000, 2))

for i in range(1000):
    random_vertex = initial_triangle[torch.randint(0, 3, (1,))]
    point = (point + random_vertex) / 2
    points_holder[i] = point


plt.figure(figsize=(4, 4))
plt.scatter(points_holder[:, 0], points_holder[:, 1], 0.15, color='blue')
plt.show()
    
```
