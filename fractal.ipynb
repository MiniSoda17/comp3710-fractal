```python
import matplotlib.pyplot as plt
import torch

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

initial_triangle = torch.tensor([[0.0, 0.0], [1.0, 0.0], [0.5, 0.866]], device=device) 

points_holder = torch.rand(50000, 2, device=device)

for i in range(30):
    random_vertex = initial_triangle[torch.randint(0, 3, (50000,), device=device)]
    points_holder = (points_holder + random_vertex) / 2

plot = points_holder.cpu().numpy()

plt.figure(figsize=(6, 6))
plt.scatter(plot[:, 0], plot[:, 1], s=0.1, color='blue')
plt.show()
```
